```python
columns_to_drop = ['player_url', 'fifa_version', 'fifa_update', 'update_as_of', 'dob', 'height_cm', 'weight_kg',
                        'club_team_id', 'league_id', 'league_level', 'club_position', 'club_jersey_number',
                        'club_loaned_from', 'club_joined_date', 'nationality_id',
                        'nation_team_id', 'nation_position', 'nation_jersey_number', 'preferred_foot', 'weak_foot', 'skill_moves',
                        'international_reputation', 'body_type', 'real_face', 'ls', 'st', 'rs', 'lw', 'lf', 'cf', 'rf', 'rw', 'lam', 'cam', 'ram', 'lm', 'lcm', 'cm', 'rcm', 'rm', 'lwb',
                        'ldm', 'cdm', 'rdm', 'rwb', 'lb', 'lcb', 'cb', 'rcb', 'rb', 'gk']



df_caracteristics = df.drop(columns=columns_to_drop, axis =1)
```

```python
columns_to_filter = ['pace', 'shooting', 'passing', 'dribbling', 'defending', 'physic']
df_nan = df_caracteristics[df_caracteristics[columns_to_filter].isna().any(axis=1)]
df_nan['player_positions'].unique()
# Making sure that all NaN values are from goalkeepers
```

```python
list_nan_gs = df_caracteristics[df_caracteristics['goalkeeping_speed'].isna()]['player_positions'].unique()

"GK" in list_nan_gs

# Makign sure there were no goalkeepers with NaN on goalkeeping_speed
```

```python
df_caracteristics = df_caracteristics.fillna(0)

df_caracteristics.isna().sum()
```

```python
df_caracteristics['work_rate'].unique()
```

```python
df_caracteristics[['attacking_work_rate', 'defending_work_rate']] = df_caracteristics['work_rate'].str.split('/', expand=True)
df_caracteristics.drop(columns='work_rate', axis=1, inplace=True)
df_caracteristics.head()
```

```python
# Para manter na escala de 0 - 100, vamos atribuir valores numéricos à essas duas colunas que podem ser traduzidas como 
# Comprometimetno no ataque e Comprometimento na defesa. Já que não há diferenciação baseado no overall, 33 será baixo comprometimento, 66 médio e 99 alto

df_caracteristics['attacking_work_rate'] = df_caracteristics['attacking_work_rate'].apply(lambda x: 33 if x == 'Low' else (66 if x == 'Medium' else 99))


df_caracteristics['defending_work_rate'] = df_caracteristics['defending_work_rate'].apply(lambda x: 33 if x == 'Low' else (66 if x == 'Medium' else 99))
```

```python
file_path = 'data\\silver.csv'

# 
data_folder = os.path.dirname(file_path)
if not os.path.exists(data_folder):
    os.makedirs(data_folder)
    print(f"Pasta criada: {data_folder}")

# Salva o DataFrame no arquivo CSV
df_caracteristics.to_csv(file_path, index=False)

print(f"Arquivo salvo: {file_path}")
```
